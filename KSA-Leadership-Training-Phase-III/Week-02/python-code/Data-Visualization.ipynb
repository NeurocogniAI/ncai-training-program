# Load & Visualize a NIfTI file (MRI/fMRI/DTI/PET)


```python
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt

# Load NIfTI (works for MRI, fMRI, DTI, PET)
img = nib.load('./data/chris_t1.nii.gz')  # structural MRI
data = img.get_fdata()
affine = img.affine

# Show middle slice
mid_slice = data.shape[2] // 2
plt.imshow(data[mid_slice, :, :].T, cmap='gray', origin='lower')
plt.title('Structural MRI (T1)')
plt.axis('off')
plt.show()

print(f"Data shape: {data.shape}")
print(f"Voxel size: {img.header.get_zooms()}")
```
